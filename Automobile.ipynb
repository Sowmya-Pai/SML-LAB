```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
d = pd.read_csv("/kaggle/input/datasets/sowmyaupai/automobile/Automobile.csv")
```

```python
d.shape
```

```python
d.columns = d.columns.str.strip()
```

```python
numeric_cols = ['Fuel efficiency (mpg)', 'Engine size (lit)', 'Price (in thousands)', 'horsepower', 'acceleration']
d_numeric = d[numeric_cols]
```

```python
sns.pairplot(d_numeric, diag_kind='kde', corner=True)
```

```python
corr = d_numeric.corr()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(corr, annot=True, cmap='plasma', fmt=".2f")
plt.title("Correlation Matrix of Car Features")
plt.show()
```
